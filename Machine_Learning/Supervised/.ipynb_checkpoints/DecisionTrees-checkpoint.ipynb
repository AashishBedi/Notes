```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
dataframe = pd.read_csv("C:/Users/91798/Desktop/DATA_ENGINEERING/Machine_Learning/Supervised/WineQT.csv")
dataframe.head()
```

```python
dataframe.columns
```

```python
set(dataframe['quality'])
```

```python
##Spliting the data into X(Independant features) AND Y(target column)
```

```python
X = dataframe.drop(columns='quality', axis=1)
y = dataframe['quality']
```

```python
##Spliting data into training and testing
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y)
```

```python
# Data Modeling
from sklearn.tree import DecisionTreeClassifier
model = DecisionTreeClassifier()
model.fit(X_train, y_train)
```

```python
#Model Predictions
y_pred = model.predict(X_test)
y_pred
```

```python
from sklearn import tree
plt.figure(figsize=(20,20))
tree.plot_tree(model, filled=True)
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_pred, y_test))
```

```python
from sklearn.ensemble import RandomForestClassifier, RandomForestRegressor
#Training the initial rf model
rf = RandomForestClassifier()
rf.fit(X_train, y_train)
```

```python
#Evaluating the Initial Model:
rf.score(X_test, y_test)
```

```python
#Making Predictions with the Initial Model:
y_predict_rf = rf.predict(X_test)
print(y_predict_rf)
```

```python
#Accessing the Individual Trees in the Forest:
rf.estimators_
```

```python
#Defining Hyperparameter Grid for Grid Search:
grid_param =  {
    "n_estimators" : [50, 100, 120],
    'criterion' : ['gini', 'entropy'],
    'max_depth' : range(5)
}
```

```python
#Performing Grid Search with Cross-Validation:
from sklearn.model_selection import GridSearchCV
grid_search_rf = GridSearchCV(param_grid=grid_param, cv=10, n_jobs=6, verbose=1, estimator=rf)
grid_search_rf.fit(X_train, y_train)
```

```python
#Retrieving the Best Hyperparameters:
grid_search_rf.best_params_
```

```python
#Training the Optimized Random Forest Model:
rf_new = RandomForestClassifier(n_estimators=120, criterion='entropy', max_depth=4)
rf_new.fit(X_train, y_train)
```

```python
#Evaluating the Optimized Model:
rf_new.score(X_test, y_test)
```
