```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv('/content/diabetes.csv')
data.head()
## descriptive statistics of a given data
data.describe()
```

```python
##3. Check the standard normal form

## input features
data = data.drop(columns=['Outcome'], axis=1)
## convert the data into standard scaler form
## mean = 0 and standard deviation = 1
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
dataScaled = scaler.fit_transform(data)
dataScaled = pd.DataFrame(dataScaled)
dataScaled.head()
```

```python
#4. Find Covariance Matrix

covarianceMatrix = dataScaled.T @ dataScaled / 767
covarianceMatrix
```

```python
#5. Evaluate the EigenValues and EigenVectors using the Covariance Matrix

eigenValues, eigenVectors = np.linalg.eig(covarianceMatrix)
print(eigenValues)
```

```python
#6. Select Principal Components

## PC1 data
PC1_data = dataScaled @ eigenVectors[:, 0]
## PC2 data
PC2_data = dataScaled @ eigenVectors[:, 1]
## PC3 data
PC3_data = dataScaled @ eigenVectors[:, 7]
```

```python
#7. PCA Implementation via the sklearn library

from sklearn.decomposition import PCA
pca = PCA(n_components=3)
pca.fit_transform(dataScaled)
```

```python
#8. Calculate the explained variance ratio

from sklearn.decomposition import PCA
pca = PCA()
principalComponent = pca.fit_transform(dataScaled)
pca.explained_variance_ratio_
```

```python
#9. Visualizing Cumulative Explained Variance in PCA

import matplotlib.pyplot as plt
plt.plot(np.cumsum(pca.explained_variance_ratio_))
plt.xlabel("Dimensions")
plt.ylabel("Explained Variance Ratio")
plt.savefig("ScreenPlot.png")
plt.show()
```
